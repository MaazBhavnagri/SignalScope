```python
# ── Cell 5: Write datasets.py (ManifestDataset + transforms) ───────────────
datasets_code = '''
from __future__ import annotations
import csv
from pathlib import Path
import numpy as np
from PIL import Image
import torch
from torch.utils.data import Dataset
import torchvision.transforms as T
from model import config as C

def load_manifest(split: str | None = None, sources: list[str] | None = None) -> list[dict]:
    if not C.MANIFEST_PATH.exists():
        return []
    rows = []
    with open(C.MANIFEST_PATH) as f:
        for r in csv.DictReader(f):
            if split and r["split"] != split:
                continue
            if sources and r["source"] not in sources:
                continue
            rows.append(r)
    return rows

def balanced_sample_weights(rows: list[dict]) -> list[float]:
    labels = [int(r["label"]) for r in rows]
    n0 = labels.count(0) or 1
    n1 = labels.count(1) or 1
    w = {0: 1.0 / n0, 1: 1.0 / n1}
    return [w[l] for l in labels]

def train_transform(img_size: int = 224) -> T.Compose:
    return T.Compose([
        T.Resize((img_size + 32, img_size + 32)),
        T.RandomCrop(img_size),
        T.RandomHorizontalFlip(),
        T.ColorJitter(0.2, 0.2, 0.1, 0.05),
        T.RandomGrayscale(p=0.05),
        T.ToTensor(),
        T.Normalize(C.IMAGENET_MEAN, C.IMAGENET_STD),
        # JPEG simulation via random quality re-encode
        T.RandomApply([T.GaussianBlur(3, sigma=(0.1, 1.5))], p=0.3),
    ])

def eval_transform(img_size: int = 224) -> T.Compose:
    return T.Compose([
        T.Resize((img_size, img_size)),
        T.ToTensor(),
        T.Normalize(C.IMAGENET_MEAN, C.IMAGENET_STD),
    ])

class ManifestDataset(Dataset):
    def __init__(self, rows: list[dict], transform):
        self.rows = rows
        self.transform = transform
        self.attr_index = C.ATTR_INDEX

    def __len__(self): return len(self.rows)

    def __getitem__(self, idx):
        r = self.rows[idx]
        img = Image.open(r["path"]).convert("RGB")
        x = self.transform(img)
        y = torch.tensor(int(r["label"]), dtype=torch.float32)
        attr = torch.tensor(self.attr_index.get(r.get("generator", ""), C.ATTR_IGNORE), dtype=torch.long)
        return x, y, attr, r["path"]
'''
with open('/content/SignalScope/model/data/datasets.py', 'w') as f:
    f.write(datasets_code)
print('datasets.py written')
```

```python
# ── Cell 6: Download & prepare dataset ────────────────────────────────────
# Uses two free HuggingFace datasets cited in your PLAN.md:
#   - CIFAKE (100k images, 32px, real CIFAR-10 + SD1.4 fakes)
#   - Tiny-GenImage (35k images, 8 generators including Midjourney/BigGAN as held-out)

import csv, os, random, shutil
from pathlib import Path
from PIL import Image
from datasets import load_dataset
from tqdm.auto import tqdm

PROCESSED = Path('/content/SignalScope/data/processed')
PROCESSED.mkdir(parents=True, exist_ok=True)
MANIFEST = Path('/content/SignalScope/data/manifest.csv')

SEED = 1337
random.seed(SEED)

# Generator -> attribution class mapping
GEN_MAP = {
    'stable_diffusion_v_1_4': 'SD14', 'stable_diffusion_v_1_5': 'SD15',
    'ADM': 'ADM', 'GLIDE': 'GLIDE', 'Midjourney': 'Midjourney',
    'VQDM': 'VQDM', 'Wukong': 'Wukong', 'BigGAN': 'BigGAN',
    'real': 'Real', 'REAL': 'Real',
}
HELD_OUT = {'Midjourney', 'VQDM'}

rows = []

def save_img(img, dest: Path):
    dest.parent.mkdir(parents=True, exist_ok=True)
    if not dest.exists():
        img.convert('RGB').save(dest, 'JPEG', quality=95)

# ── CIFAKE ──
print('Loading CIFAKE...')
cifake = load_dataset('dragonintelligence/CIFAKE-image-dataset', split='train')
cifake_val = load_dataset('dragonintelligence/CIFAKE-image-dataset', split='test')

CIFAKE_PER_CLASS = 20000
real_idx = [i for i, x in enumerate(cifake) if x['label'] == 0]
fake_idx = [i for i, x in enumerate(cifake) if x['label'] == 1]
random.shuffle(real_idx); random.shuffle(fake_idx)
selected = [(i, 'train') for i in real_idx[:CIFAKE_PER_CLASS] + fake_idx[:CIFAKE_PER_CLASS]]
val_real = [i for i, x in enumerate(cifake_val) if x['label'] == 0][:1000]
val_fake = [i for i, x in enumerate(cifake_val) if x['label'] == 1][:1000]
selected += [(i, 'val_cifake') for i in val_real + val_fake]

for idx, split_tag in tqdm(selected, desc='CIFAKE'):
    ds = cifake if split_tag == 'train' else cifake_val
    item = ds[idx]
    lbl = item['label']  # 0=real, 1=fake
    gen = 'Real' if lbl == 0 else 'SD14'
    split = split_tag if split_tag == 'train' else 'val'
    fname = PROCESSED / split / f'cifake_{idx}_{lbl}.jpg'
    save_img(item['image'], fname)
    rows.append({'path': str(fname), 'label': lbl, 'generator': gen, 'source': 'cifake', 'split': split})

print(f'CIFAKE done: {len(rows)} rows')

# ── Tiny-GenImage ──
print('Loading Tiny-GenImage...')
try:
    genimg = load_dataset('TheKernel01/Tiny-GenImage', split='train')
    gen_items = list(genimg)
    random.shuffle(gen_items)
    for i, item in enumerate(tqdm(gen_items, desc='Tiny-GenImage')):
        gen_name = GEN_MAP.get(item.get('generator', item.get('label_str', '')), None)
        if gen_name is None: continue
        lbl = 0 if gen_name == 'Real' else 1
        # assign split: held-out generators go to test only
        if gen_name in HELD_OUT:
            split = 'test'
        elif i % 10 == 0:
            split = 'test'
        elif i % 10 == 1:
            split = 'val'
        else:
            split = 'train'
        fname = PROCESSED / split / f'genimg_{i}_{lbl}.jpg'
        save_img(item['image'], fname)
        rows.append({'path': str(fname), 'label': lbl, 'generator': gen_name, 'source': 'tiny_genimage', 'split': split})
    print(f'Tiny-GenImage done, total rows now: {len(rows)}')
except Exception as e:
    print(f'Tiny-GenImage failed ({e}), continuing with CIFAKE only')

# ── Write manifest ──
with open(MANIFEST, 'w', newline='') as f:
    w = csv.DictWriter(f, fieldnames=['path','label','generator','source','split'])
    w.writeheader(); w.writerows(rows)

from collections import Counter
splits = Counter(r['split'] for r in rows)
print('Manifest written:', dict(splits))
```
